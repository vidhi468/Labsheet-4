```python
# Q1. Load the Iris dataset
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

print(df.head())
```

```python
# # Q2. Display first and last five records

print("First five records:")
print(df.head())

print("\nLast five records:")
print(df.tail())
```

```python
# # Q3. Dataset information

print("Dataset Information:")
df.info()

print("\nSummary Statistics:")
print(df.describe())

print("\nData Types:")
print(df.dtypes) 
```

```python
# # Q4. Select numerical features

from sklearn.datasets import load_iris

iris = load_iris()

X = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

print("Selected numerical features:")
print(X.head())
```

```python
# # Q5. Standardize the numerical features

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

print("Standardized data:")
print(X_scaled.head())
```

```python
# # Q6. Implement K-Means

from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

kmeans.fit(X_scaled)

print("K-Means clustering completed.")
print("Cluster labels:")
print(kmeans.labels_)
```

```python
# # Q7. K-Means with K = 2

kmeans_2 = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

labels_2 = kmeans_2.fit_predict(X_scaled)

print("Cluster labels for K=2:")
print(labels_2)
```

```python
# Q8. K-Means with K = 3

kmeans_3 = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

labels_3 = kmeans_3.fit_predict(X_scaled)

print("Cluster labels for K=3:")
print(labels_3)
```

```python
# Q9. Compare different K values

for k in range(2, 7):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    print(f"K = {k}")
    print("Cluster counts:")
    print(pd.Series(labels).value_counts().sort_index())
    print()
```

```python
# Q10. Elbow Method

import matplotlib.pyplot as plt

inertia_values = []

for k in range(1, 11):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(X_scaled)

    inertia_values.append(model.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    inertia_values,
    marker='o'
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.grid(True)
plt.show()
```

```python
# Q11. Scatter plot of K-Means clusters

plt.figure(figsize=(8, 6))

plt.scatter(
    X_scaled.iloc[:, 0],
    X_scaled.iloc[:, 1],
    c=labels_3,
    s=50
)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("K-Means Clustering (K=3)")

plt.show()
```

```python
# Q12. Display cluster centroids

centroids = kmeans_3.cluster_centers_

centroid_df = pd.DataFrame(
    centroids,
    columns=X.columns
)

print("Cluster Centroids:")
print(centroid_df)
```

```python
# Q13. Add cluster labels to original dataset

clustered_df = X.copy()

clustered_df["Cluster"] = labels_3

print(clustered_df.head(10))
```

```python
# Q14. K-Means before scaling

kmeans_unscaled = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

labels_unscaled = kmeans_unscaled.fit_predict(X)

# K-Means after scaling

kmeans_scaled = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

labels_scaled = kmeans_scaled.fit_predict(X_scaled)

print("Before scaling:")
print(pd.Series(labels_unscaled).value_counts().sort_index())

print("\nAfter scaling:")
print(pd.Series(labels_scaled).value_counts().sort_index())
```

```python
# Q15. Agglomerative Hierarchical Clustering

from sklearn.cluster import AgglomerativeClustering

hierarchical = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

hierarchical_labels = hierarchical.fit_predict(X_scaled)

print("Hierarchical cluster labels:")
print(hierarchical_labels)
```

```python
# Q16. Generate dendrogram

from scipy.cluster.hierarchy import dendrogram, linkage

linkage_matrix = linkage(
    X_scaled,
    method="ward"
)

plt.figure(figsize=(12, 6))

dendrogram(linkage_matrix)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")

plt.show()
```

```python
# Q17. Ward linkage

ward_model = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

ward_labels = ward_model.fit_predict(X_scaled)

print("Ward linkage labels:")
print(ward_labels)
```

```python
# Q18. Complete linkage

complete_model = AgglomerativeClustering(
    n_clusters=3,
    linkage="complete"
)

complete_labels = complete_model.fit_predict(X_scaled)

print("Complete linkage labels:")
print(complete_labels)
```

```python
# Q19. Compare linkage methods

linkage_methods = [
    "ward",
    "complete",
    "average",
    "single"
]

for method in linkage_methods:

    if method == "ward":
        model = AgglomerativeClustering(
            n_clusters=3,
            linkage=method
        )
    else:
        model = AgglomerativeClustering(
            n_clusters=3,
            linkage=method
        )

    labels = model.fit_predict(X_scaled)

    print(f"\nLinkage method: {method}")
    print(pd.Series(labels).value_counts().sort_index())
```

```python
# Q20. Visualize hierarchical clusters

plt.figure(figsize=(8, 6))

plt.scatter(
    X_scaled.iloc[:, 0],
    X_scaled.iloc[:, 1],
    c=ward_labels,
    s=50
)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Hierarchical Clustering - Ward Linkage")

plt.show()
```

```python
# Q21. Compare K-Means and Hierarchical Clustering

comparison_df = pd.DataFrame({
    "KMeans": labels_3,
    "Hierarchical": ward_labels
})

print(comparison_df.head(20))
```

```python
# Q22. Apply PCA

from sklearn.decomposition import PCA

pca = PCA()

X_pca = pca.fit_transform(X_scaled)

print("Original dimensions:", X_scaled.shape)
print("PCA dimensions:", X_pca.shape)
```

```python
# Q23. Reduce to two principal components

pca_2 = PCA(n_components=2)

X_pca_2 = pca_2.fit_transform(X_scaled)

pca_df = pd.DataFrame(
    X_pca_2,
    columns=["PC1", "PC2"]
)

print(pca_df.head())
```

```python
# Q24. PCA 2D scatter plot

plt.figure(figsize=(8, 6))

plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"],
    s=50
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA - Two Dimensional Representation")

plt.grid(True)
plt.show()
```

```python
# Q25. Explained variance ratio

print("Explained Variance Ratio:")

for i, variance in enumerate(
    pca.explained_variance_ratio_,
    start=1
):
    print(f"PC{i}: {variance:.4f}")
```

```python
# Q26. Cumulative explained variance

cumulative_variance = pca.explained_variance_ratio_.cumsum()

for i, value in enumerate(
    cumulative_variance,
    start=1
):
    print(
        f"First {i} components: "
        f"{value:.4f}"
    )

plt.figure(figsize=(8, 5))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker="o"
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance")

plt.grid(True)
plt.show()
```

```python
# Q27. Compare dimensions

print("Before PCA:")
print(X_scaled.shape)

print("\nAfter PCA:")
print(X_pca_2.shape)
```

```python
# Q28. K-Means on PCA data

kmeans_pca = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

pca_labels = kmeans_pca.fit_predict(X_pca_2)

pca_df["Cluster"] = pca_labels

print(pca_df.head())
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"],
    c=pca_df["Cluster"],
    s=50
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("K-Means Clustering on PCA Data")

plt.show()
```

```python
# Q29. Compare clustering performance

from sklearn.metrics import silhouette_score

score_before_pca = silhouette_score(
    X_scaled,
    labels_3
)

score_after_pca = silhouette_score(
    X_pca_2,
    pca_labels
)

print(
    "Silhouette Score before PCA:",
    round(score_before_pca, 4)
)

print(
    "Silhouette Score after PCA:",
    round(score_after_pca, 4)
)
```

```python
# Q30. Silhouette Score

silhouette = silhouette_score(
    X_scaled,
    labels_3
)

print(
    "Silhouette Score:",
    round(silhouette, 4)
)
```

```python
# Q31. Compare Silhouette Scores

from sklearn.metrics import silhouette_score

scores = []

for k in range(2, 8):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    scores.append(score)

    print(
        f"K = {k}, "
        f"Silhouette Score = {score:.4f}"
    )
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(2, 8),
    scores,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")

plt.grid(True)
plt.show()
```
